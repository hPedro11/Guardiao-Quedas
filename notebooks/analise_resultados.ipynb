## Métricas por configuração

```python
d = json.loads((RES / 'metricas.json').read_text(encoding='utf-8'))
print(f"Clipes: {d['n_clipes']} ({d['n_queda']} com queda) | modelo: {d['modelo']} | gerado em {d['gerado_em']}")
for nome, c in d['configs'].items():
    print(f"\n{nome}")
    print('  matriz:', c['matriz'])
    for k, v in c['metricas'].items():
        print(f'  {k}: {v:.3f}')
    print(f"  FPS médio: {c['fps_medio']} | quadros com pose: {c['taxa_pose_media']*100:.0f}%")
```

```python
display(Image(str(RES / 'comparacao_metricas.png')))
for nome in d['configs']:
    display(Image(str(RES / f'matriz_confusao_{nome}.png')))
```

## Resultado por clipe

```python
with open(RES / 'resultados.csv', encoding='utf-8') as f:
    linhas = list(csv.DictReader(f))
erros = [l for l in linhas if l['real_queda'] != l['previsto_queda']]
print(f'{len(erros)} erro(s) em {len(linhas)} execuções')
for l in erros:
    print(' ', l['config'], l['clipe'], '| real:', l['real_queda'], '| previsto:', l['previsto_queda'])
```

## Curvas por clipe
